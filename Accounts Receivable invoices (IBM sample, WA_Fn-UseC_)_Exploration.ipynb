## Why we chose dataset 1(Accounts Receivable Invoices(IBM sample, WA_Fn-UseC_))
Rent payments work like invoices, so this dataset is the closest match to the rent problem. It also allows time-based testing using 2012 data for training and 2013 data for testing.

**The main limitation is that it uses business invoices, not real tenant records.**
